```python
import pandas as pd  # Correct import
import numpy as np
import matplotlib.pyplot as plt
import datetime
```

```python
def max_subarray_sum(arr):
    max_sum = float('-inf')  # Initialize max_sum to negative infinity
    current_sum = 0
    start = 0
    end = 0
    temp_start = 0

    for i in range(len(arr)):
        current_sum += arr[i]
        # Update max_sum if current_sum is greater
        if current_sum > max_sum:
            max_sum = current_sum
            start = temp_start
            end = i  # Update end index
   # If current_sum becomes negative, reset it and update temp_start
        if current_sum < 0:
            current_sum = 0
            temp_start = i + 1
  # Return the maximum subarray sum and the subarray itself
    return max_sum, arr[start : end + 1]
```

```python
 # Given array
n = [-3, -1, 1, -4, 3, 5, -2, 6]

# Finding the maximum subarray
current_time = datetime.datetime.now().time()
max_sum, subarray = max_subarray_sum(n)
print("Maximum Sum:", max_sum)
print("Subarray:", subarray)
```

```python
def max_subarray_sum(arr):
    max_sum = current_sum = arr[0]
    start = end = s = 0

    for i in range(1, len(arr)):
        if arr[i] > current_sum + arr[i]:
            current_sum = arr[i]
            s = i
        else:
            current_sum += arr[i]

        if current_sum > max_sum:
            max_sum = current_sum
            start = s
            end = i

    return max_sum, arr[start:end + 1]
```

```python
# Worst-case: All negative elements
n = [-3, -2, -4, -1]

# Get the result
max_sum, subarray = max_subarray_sum(n)

# Output the result
print("Maximum Sum:", max_sum)
print("Subarray:", subarray)
```
