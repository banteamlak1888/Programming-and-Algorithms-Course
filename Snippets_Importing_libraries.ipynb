```
!pip install python-docx
from docx import Document
from docx.shared import Inches, Pt
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_CELL_VERTICAL_ALIGNMENT
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

out = "/content/Embedded_Systems_Course_Outline.docx"

doc = Document()
sec = doc.sections[0]
sec.top_margin = Inches(0.7)
sec.bottom_margin = Inches(0.7)
sec.left_margin = Inches(0.8)
sec.right_margin = Inches(0.8)

styles = doc.styles
styles["Normal"].font.name = "Times New Roman"
styles["Normal"].font.size = Pt(11)

def set_cell_text(cell, text, bold=False):
    cell.text = ""
    p = cell.paragraphs[0]
    r = p.add_run(text)
    r.bold = bold
    r.font.name = "Times New Roman"
    r.font.size = Pt(10)
    cell.vertical_alignment = WD_CELL_VERTICAL_ALIGNMENT.CENTER

def shade_cell(cell, fill="D9EAF7"):
    tcPr = cell._tc.get_or_add_tcPr()
    shd = OxmlElement("w:shd")
    shd.set(qn("w:fill"), fill)
    tcPr.append(shd)

# Header
p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
r = p.add_run("MEKDELA AMBA UNIVERSITY")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(14)

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
r = p.add_run("BACHELOR OF SCIENCE DEGREE\nIN\nSOFTWARE ENGINEERING")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)

# Course information
info = [
    ("Course Title", "Embedded Systems"),
    ("Course Code", "SEng5188"),
    ("CP", "5 (2hr Lecture, 3hr Tutorial)"),
    ("Module Title", "Electives"),
    ("Module Number", "18"),
    ("Pre-requisites", "None"),
    ("Year", "V"),
    ("Semester", "II"),
    ("Status of Course", "Elective II"),
]
table = doc.add_table(rows=0, cols=2)
table.alignment = WD_TABLE_ALIGNMENT.CENTER
table.style = "Table Grid"
for k, v in info:
    cells = table.add_row().cells
    set_cell_text(cells[0], k, True)
    set_cell_text(cells[1], v)
    cells[0].width = Inches(1.7)
    cells[1].width = Inches(4.8)

doc.add_paragraph()

# Description
p = doc.add_paragraph()
r = p.add_run("Course Description")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)

desc = (
    "This course introduces the principles, components, design process, and implementation of "
    "embedded computer systems. It focuses on the integration of embedded hardware and software "
    "to meet functional, performance, reliability, and real-time requirements."
)
p = doc.add_paragraph(desc)
p.paragraph_format.space_after = Pt(6)

# Objectives
p = doc.add_paragraph()
r = p.add_run("Course Objectives")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)

objectives = [
    "Understand the fundamental concepts and characteristics of embedded systems.",
    "Understand basic embedded system hardware and software components.",
    "Apply basic principles of embedded system design and development.",
    "Understand sensors, actuators, input/output, and communication interfaces.",
    "Understand basic real-time, reliability, security, and safety considerations.",
    "Develop and demonstrate a simple embedded system prototype."
]
for item in objectives:
    p = doc.add_paragraph(style=None)
    p.style = doc.styles["Normal"]
    p.paragraph_format.left_indent = Inches(0.2)
    p.add_run("➢ " + item)

# Course contents
p = doc.add_paragraph()
r = p.add_run("Course Contents")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)

rows = [
    ("1–3", "Chapter 1: Fundamentals of Embedded Systems",
     "Introduction to embedded systems\nBasic embedded system architecture\nCharacteristics and types of embedded systems\nEmbedded system requirements\nEmbedded system design process",
     "Assignment 1: Embedded System Analysis"),
    ("4–6", "Chapter 2: Embedded Software and Programming",
     "Embedded software and firmware\nProgramming concepts for embedded systems\nSoftware development tools\nDevelopment and debugging process\nIntroduction to real-time embedded systems",
     "Assignment 2: Embedded Software and Development Tools"),
    ("7–9", "Chapter 3: Embedded Hardware and Microcontrollers",
     "Embedded hardware components\nMicroprocessors and microcontrollers\nMemory and input/output\nMicrocontroller architecture\nDevelopment boards and basic hardware selection",
     "Assignment 3: Microcontroller and Development Board Comparison"),
    ("10–12", "Chapter 4: Sensors, Actuators and Communication",
     "Sensors and actuators\nInput/output interfacing\nBasic sensor and actuator integration\nCommunication interfaces\nEmbedded communication systems",
     "Assignment 4: Sensor–Actuator System Design"),
    ("13–16", "Chapter 5: Embedded System Design and Project",
     "Hardware and software integration\nEmbedded system testing and debugging\nReliability, security and safety considerations\nEmbedded system case studies\nProject development and presentation",
     "Course Project: Design, Implementation, Testing and Demonstration"),
]

t = doc.add_table(rows=1, cols=4)
t.style = "Table Grid"
t.alignment = WD_TABLE_ALIGNMENT.CENTER
headers = ["Weeks", "Lecture Topics", "Main Contents", "Reading / Assignments / Activities"]
for i, h in enumerate(headers):
    set_cell_text(t.rows[0].cells[i], h, True)
    shade_cell(t.rows[0].cells[i])

for row in rows:
    cells = t.add_row().cells
    for i, val in enumerate(row):
        set_cell_text(cells[i], val)
    # bold chapter title
    cells[1].paragraphs[0].runs[0].bold = True

# Teaching methods
p = doc.add_paragraph()
r = p.add_run("Summary of Teaching Learning Methods")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)
p = doc.add_paragraph(
    "The course will be delivered through lectures, demonstrations, tutorials, practical activities, "
    "group discussions, assignments, student presentations, and an individual or group project."
)

# Assessment
p = doc.add_paragraph()
r = p.add_run("Assessment Methods")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)
assessment = [
    ("Assignments", "20%"),
    ("Mid Examination", "20%"),
    ("Final Examination", "30%"),
    ("Project", "30%"),
    ("Total", "100%"),
]
at = doc.add_table(rows=1, cols=2)
at.style = "Table Grid"
at.alignment = WD_TABLE_ALIGNMENT.CENTER
set_cell_text(at.rows[0].cells[0], "Assessment", True)
set_cell_text(at.rows[0].cells[1], "Weight", True)
shade_cell(at.rows[0].cells[0]); shade_cell(at.rows[0].cells[1])
for a, w in assessment:
    cells = at.add_row().cells
    set_cell_text(cells[0], a, a == "Total")
    set_cell_text(cells[1], w, a == "Total")

# Project requirements
p = doc.add_paragraph()
r = p.add_run("Course Project")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)
for item in [
    "Project proposal and system requirements.",
    "System architecture and design.",
    "Hardware and software implementation.",
    "Testing and debugging.",
    "Technical report and presentation.",
    "Functional demonstration of the prototype."
]:
    p = doc.add_paragraph("➢ " + item)

# References - concise, based on supplied curriculum
p = doc.add_paragraph()
r = p.add_run("References")
r.bold = True; r.font.name = "Times New Roman"; r.font.size = Pt(12)
refs = [
    'Michael Barr, "Embedded Systems Glossary," Neutrino Technical Library.',
    'Steve Heath, Embedded Systems Design, 2nd ed., Newnes, 2003.',
    'Michael Barr and Anthony J. Massa, Programming Embedded Systems with C and GNU Development Tools, O’Reilly, 2006.',
    'C. Alippi, Intelligence for Embedded Systems, Springer, 2014.',
    'S. Mittal, "A Survey of Techniques for Improving Energy Efficiency in Embedded Computing Systems," IJCAET, 2014.'
]
for ref in refs:
    doc.add_paragraph("➢ " + ref)

doc.save(out)
print(f"Created: {out}")
```
